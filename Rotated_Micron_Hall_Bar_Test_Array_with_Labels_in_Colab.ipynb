```python
"""This script generates a parameterized Hall bar array using PHIDL and GDSTK."""

from dataclasses import dataclass
from pathlib import Path
from typing import Final

import gdstk
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.collections import PolyCollection
from matplotlib.patches import Patch

try:
    from phidl import Device
except ImportError:
    Device = None


# ---------------------------------------------------------------------------
# Control knobs
# ---------------------------------------------------------------------------

OUTPUT_DIR: Final[Path] = Path("/content")
OUTPUT_BASENAME: Final[str] = "Rotated_Hall_bar_array"
EXPORT_FORMAT: Final[str] = "GDS"

# Set the desired array size and Hall bar main trace width here.
# The default Hall bar trace width is set to 10 microns.
ARRAY_COLUMNS_X: Final[int] = 12
ARRAY_ROWS_Y: Final[int] = 8
MAIN_HALL_BAR_TRACE_WIDTH_UM: Final[float] = 10.0

# The incremental rotation angle applied successively to each array element.
ROTATION_INCREMENT_CW_DEG: Final[float] = 15.0

X_EDGE_TO_EDGE_SPACING_UM: Final[float] = 285.0
Y_EDGE_TO_EDGE_SPACING_UM: Final[float] = 480.0

LABEL_TOP_OFFSET_UM: Final[float] = 150.0
LABEL_LEFT_INSET_UM: Final[float] = 5.0
LABEL_HEIGHT_UM: Final[float] = 120.0

GDS_UNIT_M: Final[float] = 1e-6
GDS_PRECISION_M: Final[float] = 1e-9

PREVIEW_DPI: Final[int] = 500
PREVIEW_ALPHA: Final[float] = 0.60
PREVIEW_LINEWIDTH: Final[float] = 0.45

plt.rcParams["figure.dpi"] = PREVIEW_DPI
plt.rcParams["font.sans-serif"] = ["Tahoma", "DejaVu Sans"]
plt.rcParams["font.weight"] = "normal"

ELECTRODE_LAYER: Final[int] = 1
HALL_BAR_LAYER: Final[int] = 2
LABEL_LAYER: Final[int] = 10
DATATYPE: Final[int] = 0


# ---------------------------------------------------------------------------
# Base Geometry
# ---------------------------------------------------------------------------

BASE_ELECTRODE_POLYGONS_UM: Final[tuple[tuple[tuple[float, float], ...], ...]] = (
    (
        (207.0, 520.0), (231.0, 280.0), (201.0, 280.0),
        (152.0, 352.0), (0.0, 370.0), (0.0, 520.0),
    ),
    (
        (207.0, 0.0), (231.0, 240.0), (201.0, 240.0),
        (152.0, 168.0), (0.0, 150.0), (0.0, 0.0),
    ),
    (
        (267.0, 520.0), (243.0, 280.0), (273.0, 280.0),
        (322.0, 352.0), (474.0, 370.0), (474.0, 520.0),
    ),
    (
        (267.0, 0.0), (243.0, 240.0), (273.0, 240.0),
        (322.0, 168.0), (474.0, 150.0), (474.0, 0.0),
    ),
    (
        (150.0, 335.0), (150.0, 305.0), (190.0, 270.0),
        (198.0, 270.0), (198.0, 250.0), (190.0, 250.0),
        (150.0, 215.0), (150.0, 185.0), (0.0, 185.0),
        (0.0, 335.0),
    ),
    (
        (324.0, 335.0), (324.0, 305.0), (284.0, 270.0),
        (276.0, 270.0), (276.0, 250.0), (284.0, 250.0),
        (324.0, 215.0), (324.0, 185.0), (474.0, 185.0),
        (474.0, 335.0),
    ),
)


@dataclass(frozen=True)
class BaseGeometry:
    """Container for the fundamental electrode polygon geometry."""

    electrode_polygons: list[np.ndarray]
    electrode_bbox: tuple[float, float, float, float]


@dataclass(frozen=True)
class LayerStyle:
    """Plotting metadata for a Graphic Database System layer."""

    layer: int
    datatype: int
    name: str


LAYER_STYLES: Final[tuple[LayerStyle, ...]] = (
    LayerStyle(ELECTRODE_LAYER, DATATYPE, "Electrodes"),
    LayerStyle(HALL_BAR_LAYER, DATATYPE, "Hall bars"),
    LayerStyle(LABEL_LAYER, DATATYPE, "Labels"),
)


def get_electrode_geometry() -> BaseGeometry:
    """Return normalized electrode polygons extracted from the base coordinates."""
    polygons = [
        np.array(points, dtype=float)
        for points in BASE_ELECTRODE_POLYGONS_UM
    ]
    return BaseGeometry(polygons, (0.0, 0.0, 474.0, 520.0))


def make_hall_bar_polygons(trace_width_um: float) -> list[gdstk.Polygon]:
    """Create the Hall bar conductor from the reference dimensions."""
    if trace_width_um <= 0:
        raise ValueError("MAIN_HALL_BAR_TRACE_WIDTH_UM must be positive.")

    center_y = 260.0
    half_width = trace_width_um / 2.0

    rectangles = [
        gdstk.rectangle(
            (181.5, center_y - half_width),
            (292.5, center_y + half_width),
        ),
        gdstk.rectangle((214.5, 226.0), (219.5, 294.0)),
        gdstk.rectangle((254.5, 226.0), (259.5, 294.0)),
    ]

    merged = gdstk.boolean(rectangles, [], "or", precision=1e-3) or []

    for polygon in merged:
        polygon.layer = HALL_BAR_LAYER
        polygon.datatype = DATATYPE

    return merged


def alphanumeric_index(index: int) -> str:
    """Return alphabetical style labels for nonnegative indices."""
    if index < 0:
        raise ValueError("Index labels must be nonnegative.")

    if index < 10:
        return str(index)

    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    index -= 10
    characters: list[str] = []

    while True:
        characters.append(alphabet[index % len(alphabet)])
        index = index // len(alphabet) - 1

        if index < 0:
            break

    return "".join(reversed(characters))


def make_device_label(row_index: int, column_index: int) -> str:
    """Format the cell label using logical coordinate mapping."""
    row_label = alphanumeric_index(row_index)
    column_label = alphanumeric_index(column_index)
    return f"^{row_label}   >{column_label}"


def polygon_bbox(polygons: list[gdstk.Polygon]) -> tuple[float, float, float, float]:
    """Return the bounding box of a list of polygonal objects."""
    if not polygons:
        raise ValueError("Cannot calculate a bounding box for an empty polygon list.")

    points = np.vstack([polygon.points for polygon in polygons])
    min_x, min_y = np.min(points, axis=0)
    max_x, max_y = np.max(points, axis=0)

    return float(min_x), float(min_y), float(max_x), float(max_y)


def make_text_polygons(
    text: str,
    left_x: float,
    top_y: float,
    height_um: float,
) -> list[gdstk.Polygon]:
    """Create label text polygons with precise spatial alignment."""
    text_polygons = gdstk.text(
        text,
        height_um,
        (0.0, 0.0),
        layer=LABEL_LAYER,
        datatype=DATATYPE,
    )

    min_x, _, _, max_y = polygon_bbox(text_polygons)
    delta_x = left_x - min_x
    delta_y = top_y - max_y

    for polygon in text_polygons:
        polygon.translate(delta_x, delta_y)

    return text_polygons


def make_unit_cell(
    library: gdstk.Library,
    geometry: BaseGeometry,
    trace_width_um: float,
) -> gdstk.Cell:
    """Create one fixed electrode Hall bar unit cell."""
    unit_cell = library.new_cell("HALL_BAR_UNIT")

    for electrode in geometry.electrode_polygons:
        unit_cell.add(
            gdstk.Polygon(
                electrode,
                layer=ELECTRODE_LAYER,
                datatype=DATATYPE,
            )
        )

    for hall_bar in make_hall_bar_polygons(trace_width_um):
        unit_cell.add(hall_bar)

    return unit_cell


def make_hall_bar_array() -> tuple[gdstk.Library, gdstk.Cell, gdstk.Cell, dict[str, float]]:
    """Build the full array utilizing incremental center point rotation."""
    geometry = get_electrode_geometry()

    min_x, min_y, max_x, max_y = geometry.electrode_bbox
    electrode_width = max_x - min_x
    electrode_height = max_y - min_y

    pitch_x = electrode_width + X_EDGE_TO_EDGE_SPACING_UM
    pitch_y = electrode_height + Y_EDGE_TO_EDGE_SPACING_UM

    library = gdstk.Library(unit=GDS_UNIT_M, precision=GDS_PRECISION_M)
    unit_cell = make_unit_cell(
        library,
        geometry,
        MAIN_HALL_BAR_TRACE_WIDTH_UM,
    )
    top_cell = library.new_cell("TOP_HALL_BAR_ARRAY")

    # The intrinsic geometrical center of the unrotated unit cell bounding box.
    center_local = np.array([electrode_width / 2.0, electrode_height / 2.0], dtype=float)
    cumulative_device_index = 0

    for row in range(ARRAY_ROWS_Y):
        for column in range(ARRAY_COLUMNS_X):

            # Formulate the rotational matrices for the current incremental angle.
            rotation_deg = cumulative_device_index * ROTATION_INCREMENT_CW_DEG
            rotation_rad = -np.radians(rotation_deg)

            cosine_theta = np.cos(rotation_rad)
            sine_theta = np.sin(rotation_rad)
            rotation_matrix = np.array(
                [[cosine_theta, -sine_theta], [sine_theta, cosine_theta]],
                dtype=float
            )

            center_rotated = rotation_matrix @ center_local
            bottom_left_desired = np.array([column * pitch_x, row * pitch_y], dtype=float)
            center_desired = bottom_left_desired + center_local

            # Deduce the modified placement origin that correctly pivots the geometry around its true center.
            placement_origin = center_desired - center_rotated

            top_cell.add(
                gdstk.Reference(
                    unit_cell,
                    origin=placement_origin,
                    rotation=rotation_rad
                )
            )

            # Labels securely bypass the rotational logic to guarantee persistent horizontal legibility.
            label_text = make_device_label(row, column)
            label_left_x = bottom_left_desired[0] + LABEL_LEFT_INSET_UM
            label_top_y = bottom_left_desired[1] + min_y - LABEL_TOP_OFFSET_UM

            top_cell.add(
                *make_text_polygons(
                    label_text,
                    label_left_x,
                    label_top_y,
                    LABEL_HEIGHT_UM,
                )
            )

            cumulative_device_index += 1

    metadata = {
        "electrode_width_um": electrode_width,
        "electrode_height_um": electrode_height,
        "pitch_x_um": pitch_x,
        "pitch_y_um": pitch_y,
    }

    return library, top_cell, unit_cell, metadata


def export_layout(
    library: gdstk.Library,
    basename: str,
    export_format: str,
) -> Path:
    """Export the finalized layout architecture to the designated directory."""
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    normalized_format = export_format.strip().upper()

    if normalized_format in {"GDS", "GDSII"}:
        output_path = OUTPUT_DIR / f"{basename}.gds"
        library.write_gds(str(output_path))
        return output_path

    if normalized_format in {"OASIS", "OAS"}:
        output_path = OUTPUT_DIR / f"{basename}.oas"
        library.write_oas(str(output_path))
        return output_path

    raise ValueError('EXPORT_FORMAT must be "GDS" or "OASIS".')


def gdstk_cell_to_phidl_device(cell: gdstk.Cell):
    """Mirror a flattened GDSTK cell into a PHIDL Device object for visual inspection."""
    if Device is None:
        return None

    device = Device(f"PHIDL_PREVIEW_{cell.name}")

    for style in LAYER_STYLES:
        polygons = cell.get_polygons(
            layer=style.layer,
            datatype=style.datatype,
        )

        for polygon in polygons:
            device.add_polygon(
                polygon.points,
                layer=(style.layer, style.datatype),
            )

    return device


def plot_layout_preview(cell: gdstk.Cell, phidl_device=None, title: str = "Layout Preview") -> None:
    """Render a macroscopic preview with the descriptive legend outside the active axes."""
    fig, ax = plt.subplots(figsize=(12, 8))
    color_cycle = plt.rcParams["axes.prop_cycle"].by_key()["color"]
    handles: list[Patch] = []

    if phidl_device is not None:
        polygons_by_spec = phidl_device.get_polygons(by_spec=True)
    else:
        polygons_by_spec = {
            (style.layer, style.datatype): [
                polygon.points
                for polygon in cell.get_polygons(
                    layer=style.layer,
                    datatype=style.datatype,
                )
            ]
            for style in LAYER_STYLES
        }

    for index, style in enumerate(LAYER_STYLES):
        polygons = polygons_by_spec.get((style.layer, style.datatype), [])

        if not polygons:
            continue

        facecolor = color_cycle[index % len(color_cycle)]
        collection = PolyCollection(
            polygons,
            closed=True,
            facecolors=facecolor,
            edgecolors="black",
            linewidths=PREVIEW_LINEWIDTH,
            alpha=PREVIEW_ALPHA,
            label=style.name,
        )

        ax.add_collection(collection)
        handles.append(
            Patch(
                facecolor=facecolor,
                edgecolor="black",
                alpha=PREVIEW_ALPHA,
                label=style.name,
            )
        )

    ax.autoscale_view()
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlabel("x (µm)")
    ax.set_ylabel("y (µm)")
    ax.set_title(title)
    ax.grid(True, linewidth=0.3, alpha=0.35)

    if handles:
        ax.legend(
            handles=handles,
            loc="upper left",
            bbox_to_anchor=(1.02, 1.0),
            borderaxespad=0.0,
            frameon=True,
        )

    fig.tight_layout(rect=(0.0, 0.0, 0.82, 1.0))
    plt.show()


# Execution and mathematical generation process
library, top_cell, unit_cell, metadata = make_hall_bar_array()
output_path = export_layout(library, OUTPUT_BASENAME, EXPORT_FORMAT)

phidl_preview_unit = gdstk_cell_to_phidl_device(unit_cell)
plot_layout_preview(unit_cell, phidl_preview_unit, title="Isolated Hall Bar Unit Cell Close Up")

phidl_preview_top = gdstk_cell_to_phidl_device(top_cell)
plot_layout_preview(top_cell, phidl_preview_top, title="Parameterized Hall Bar Array Preview")

print(f"Wrote file out to {output_path}")
print(
    f"Electrode bounding box dimensions are {metadata['electrode_width_um']:.3f} µm by "
    f"{metadata['electrode_height_um']:.3f} µm"
)
print(
    f"Array pitch spacing is {metadata['pitch_x_um']:.3f} µm by "
    f"{metadata['pitch_y_um']:.3f} µm"
)
print(f"Total array layout size comprises {ARRAY_COLUMNS_X} columns by {ARRAY_ROWS_Y} rows")
print(f"Main Hall bar active trace width utilizes {MAIN_HALL_BAR_TRACE_WIDTH_UM:.3f} µm")
```
